# Hyytiälä: view and reproduce the figures

This notebook presents **130 distinct scientific figures**: 7 current main-manuscript figures, 10 current supporting figures and 113 archived historical figures. The packaged supplementary information has no embedded figures.

The normal workflow renders the current 17 from supplied processed observations and saved derived tables. Saved uncertainty intervals are **not** newly calculated. Figure 1 is fixed artwork. Historical images preserve their original cohorts, methods and geometry conventions; the current analysis does not recalculate them.

Open the gallery to browse every figure and download PNG/JPEG and available PDF/SVG versions, or download the complete ZIP. Binder sessions are temporary: download anything you want to keep before leaving.


```python
from pathlib import Path
from datetime import datetime, timezone
from IPython.display import display, FileLink
from urllib.parse import quote
import os
from release_io import prepare_release
from reviewer import render_saved, run_full, notebook_gallery

prepared = prepare_release(Path("binder_release.json"), Path("runtime"))
stamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")
output = Path("outputs") / ("saved_" + stamp)
print("Verified matching code, data and additional-figure archives.")
def file_url(path):
    prefix = os.environ.get("JUPYTERHUB_SERVICE_PREFIX", "/").rstrip("/")
    return prefix + "/files/" + quote(Path(path).as_posix())
```

```python
result = render_saved(prepared, output)
gallery = output / "gallery" / "index.html"
print(f"Ready: {len(result['figures'])} distinct figures")
display(FileLink(str(output / "all_figures.zip"), url_prefix=os.environ.get("JUPYTERHUB_SERVICE_PREFIX", "/").rstrip("/") + "/files/", result_html_prefix="Download all figures: "))
display({"text/html": notebook_gallery(gallery, file_url(gallery.parent) + "/")}, raw=True)
```

## Optional: recalculate the current analyses

The following cell is disabled by default, so **Run All skips the full calculation**. To start it deliberately, change `RUN_FULL_ANALYSIS` to `True` in that cell and run it. It uses the original production defaults (5,000 bootstrap draws and 2,000 year-block draws), all declared methods, scalar/size bootstrap and the numerical reference checks. Allow roughly an hour based on the earlier Windows run; Binder resources may prevent completion. Progress appears below and full logs remain in the new output directory.

The resulting gallery still includes the archived historical figures, explicitly labelled as archived. Download results before the session ends. Each run creates a new directory and preserves earlier results.


```python
RUN_FULL_ANALYSIS = False
if RUN_FULL_ANALYSIS:
    full_stamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")
    full_output = Path("outputs") / ("production_" + full_stamp)
    full_result = run_full(prepared, full_output, confirmed=True)
    display({"text/html": notebook_gallery(full_output / "gallery" / "index.html", file_url(full_output / "gallery") + "/")}, raw=True)
    display(FileLink(str(full_output / "all_figures.zip"), url_prefix=os.environ.get("JUPYTERHUB_SERVICE_PREFIX", "/").rstrip("/") + "/files/"))
else:
    print("Full production calculation skipped. All figures are available in the saved-input gallery above.")
```
